```python
from ultralytics import YOLO

model = YOLO("yolo11s.pt")
```

```python
data_path = "/content/Nivel-de-Cibismo-4/data.yaml"
```

```python
results = model.train(
    data=data_path,
    epochs=150,
    batch=32,
    imgsz=640,
    # Regularización
    dropout=0.1,
    # Optimizador
    optimizer="Adam",   # mejor que SGD para datasets pequeños
    lr0=0.001,        # learning rate inicial
    patience=20,
    save_period=10,
    name="chicles_v3"
)
```
